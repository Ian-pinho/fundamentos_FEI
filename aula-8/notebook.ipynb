```python
numero = int(input("Digite um número ou um negativo para encerrar: "))
intervs = [0, 0, 0, 0]
while numero>=0:
    if numero <=25:
        intervs[0]+=1
    elif numero <=50:
        intervs[1]+=1
    elif numero <=75:
        intervs[2]+=1
    elif numero <=100:
        intervs[3]+=1
    numero = int(input("Digite um número ou um negativo para encerrar: "))
print(intervs)
```
