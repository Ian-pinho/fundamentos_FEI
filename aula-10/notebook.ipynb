```python
txt = input("Escreva uma frase: ")
txt_convertido = ""
for letra in txt:
    if letra in "aeiou":
        txt_convertido = txt_convertido + chr(ord(letra) - 32)
    else:
        txt_convertido = txt_convertido + letra

print("Texto convertido:", txt_convertido)
```

```python
txt = input("Escreva uma frase: ")
txt_convertido = ""
for letra in txt:
    if ord(letra) >= 65 and ord(letra) <=90:
        txt_convertido = txt_convertido + chr(ord(letra) + 32)
    elif ord(letra) >= 97 and ord(letra) <=122:
        txt_convertido = txt_convertido + letra
    else:
        txt_convertido = txt_convertido + ""

print("Texto convertido:", txt_convertido)

if txt_convertido == txt_convertido[::-1]:
    print("A string é um palíndromo.")
else:
    print("A string não é um palíndromo.")
```
