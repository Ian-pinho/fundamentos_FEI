```python
dicionario = {
    "alpha" : 1,
    "bravo" : 2,
    "charlie" : 1,
    "delta" : 3,
    "echo" : 1
}
def procuraChave(dicionario, valores):
    chaves = []
    for chave, valor in dicionario.items():
        if valor == valores:
            chaves.append(chave)
    return(chaves)
print(dicionario, "\n")

print(procuraChave(dicionario, 1))
```

```python
from random import randint
numeros = {}
for i in range(100):
    n = randint(0, 20)
    if n not in numeros:
        numeros[n] = 1
    else:
        numeros[n]+=1
        
print(list(numeros.items()))
```
